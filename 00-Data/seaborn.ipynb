```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(
    context='notebook',
    style='darkgrid',
    palette='colorblind',
    font='sans-serif',
    font_scale=0.9
  )
```

```python
df = pd.read_csv("catalog_marketing.csv")
```

```python
df.head()
```

# Plotting univariate distributions with distplot

```python
sns.displot(data=df, x='Salary', height=3, aspect=2.5)
#height=3, width=2.5 times larger than height
```

```python
sns.displot(data=df, x='Salary', height=3, aspect=2.5, kde=True)
#height=3, width=2.5 times larger than height
```

```python
sns.displot(data=df, x='Salary', binwidth=5_000, height=3, aspect=2.5)
```

```python
sns.displot(data=df, x='Children', discrete=True, height=3, aspect=1, shrink=0.8)
```

```python
sns.displot(data=df, x='Region', discrete=True, color='orange', height=3, aspect=1.5, shrink=0.8);
```

```python
sns.displot(data=df, y='Region', discrete=True, color='salmon', height=3, aspect=1.5, shrink=0.8);
```

```python
sns.displot(data=df, x='Salary', hue='Married', binwidth=5_000, height=3, aspect=2.5)
```

```python
sns.displot(data=df, x='Salary', col="Married", hue="Married", binwidth=5_000, height=3, aspect=2.5)
```

```python
sns.displot(data=df, x='Salary', col="Region", hue="Region", binwidth=5_000, height=4, aspect=2)
```

# Visualize Categorical Data

```python
sns.barplot(data=df, y='Salary', x='Married');
```

```python
sns.catplot(data=df, x="Children", y="Amount Spent", kind="box");
```

```python
sns.catplot(data=df, x="Own Home", y="Amount Spent", hue="Gender", kind="box");
```

```python
sns.catplot(data=df, x="Married", y="Salary", hue="Region", kind="bar");
```

# Visualize Relationships with Scatterplots

```python
sns.relplot(data=df, x='Salary', y='Amount Spent', height=4, aspect=2);
```

```python
sns.relplot(data=df, x='Salary', y='Amount Spent', size="Children", height=4, aspect=2);
```

```python
sns.relplot(data=df, x='Salary', y='Amount Spent', hue="Gender", height=4, aspect=2);
```

```python
sns.relplot(data=df, x='Salary', y='Amount Spent', hue="Children", height=4, aspect=2);
```

```python
sns.relplot(data=df, x='Salary', y='Amount Spent', col="Children", hue="Children", height=4, aspect=2, palette="pastel");
```

# Quantifying Relationships with Linear Model Plots

```python
sns.lmplot(data=df, x='Salary', y='Amount Spent', col="Children", height=4, aspect=2, palette="pastel");
```

```python
sns.lmplot(data=df, x='Salary', y='Amount Spent', col="Region", hue="Region", col_wrap=2, height=4, aspect=2, palette="deep");
```
