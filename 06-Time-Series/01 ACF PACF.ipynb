# Interpreting ACF and PACF Plots for Time Series Analysis
---

The Autocorrelation Function (ACF) and Partial Autocorrelation Function (PACF) plots are the primary tools used to identify the order (p and q) of an ARIMA (AutoRegressive Integrated Moving Average) model, specifically focusing on the AR (AutoRegressive) and MA (Moving Average) components.

## 1. The Foundational Intuitive Idea: Correlation Over Time
---

> Before diving into the complex terms, think of a time series ($Y_t$) as a sequence of data points. <br>We are trying to understand the relationship between the current value ($Y_t$) and its past values ($Y_{t-1}, Y_{t-2}, \dots$).

<br>

## A. Autocorrelation Function (ACF)
---

What it is: The ACF measures the total correlation between a time series and its lagged self, at different lags.

Lag 1: Correlation between $Y_t$ and $Y_{t-1}$.

Lag 2: Correlation between $Y_t$ and $Y_{t-2}$.

...and so on.

**Why it works (The Contamination Problem):**

> Imagine you are measuring the total correlation between today's temperature ($Y_t$) and the temperature three days ago ($Y_{t-3}$). This correlation is likely high, but a large part of that correlation comes indirectly from the fact that $Y_{t-3}$ affects $Y_{t-2}$, and $Y_{t-2}$ affects $Y_{t-1}$, and $Y_{t-1}$ affects $Y_t$. The ACF captures this total, direct and indirect relationship.

**Intuition**: How strongly does a past value, *including all its downstream effects*, influence the current value?

<br>

## B. Partial Autocorrelation Function (PACF)
---

What it is: The PACF measures the pure, direct correlation between a time series and its lagged self, after removing the influence of all the intermediate lags.

Lag 3 PACF: Correlation between $Y_t$ and $Y_{t-3}$, controlling for the effects of $Y_{t-1}$ and $Y_{t-2}$.

**Why it works (The Isolation Solution):**

> The PACF is designed to isolate the direct linear impact. If $Y_{t-3}$ only influences $Y_t$ indirectly through $Y_{t-1}$ and $Y_{t-2}$, the PACF at Lag 3 will be close to zero.

**Intuition**: How strongly does a past value uniquely and directly influence the current value, independent of everything in between?

<br>

<big>

# 2. Interpretation Guide: Cut-offs vs. Decay

The key to interpreting these plots is recognizing two characteristic behaviors: Cut-off and Exponential Decay (Tapering).

The Significance of the Cut-off

A cut-off occurs when the correlation bars on the plot suddenly drop to zero (or become non-significant) after a specific lag, say $k$. The bars beyond $k$ should stay within the blue/shaded significance region. This means the time series' memory stops abruptly after $k$ periods. This abrupt stop is the most important signal for identifying the model order.

| Model Type | ACF Plot Pattern |PACF Plot Pattern |Interpretation |
|---|---|---|---|
|AR($p$) | Tapers/Decays exponentially, often oscillating. | Cuts off after lag $p$. | The current value is directly influenced by the last $p$ observations. The ACF shows the ripple effect (decay), while the PACF isolates the $p$ direct inputs (cut-off). |
|MA($q$) | Cuts off after lag $q$. | Tapers/Decays exponentially, often oscillating. | The current value is directly influenced by the last $q$ error terms (shocks). The ACF isolates the $q$ direct error inputs (cut-off), while the PACF shows the ripple effect of these errors (decay).|
|ARMA($p, q$) | Tapers/Decays exponentially or sinusoidally. | Tapers/Decays exponentially or sinusoidally. | Both the ACF and PACF show decay. This means the process is a mixture of AR and MA components, and the plots cannot cleanly isolate $p$ and $q$. Other methods must be used (e.g., AIC, BIC).


<big>

## Detailed Example

**If the PACF cuts off after Lag 2, and the ACF slowly decays:** AR(2) model is appropriate.

Reasoning: The PACF tells us that $Y_t$ is directly influenced by $Y_{t-1}$ and $Y_{t-2}$, but not $Y_{t-3}$ (after removing $Y_{t-1}$ and $Y_{t-2}$'s effects). The ACF shows the decay of this initial AR impact.

**If the ACF cuts off after Lag 1, and the PACF slowly decays:** MA(1) model is appropriate.

Reasoning: The ACF tells us that the current value is correlated with only the past 1 error term (shock), $\epsilon_{t-1}$, and this correlation immediately stops. The PACF shows the gradually diminishing indirect effect of that $\epsilon_{t-1}$ over time.

<br>

# 3. Why Cut-offs Select the Model Order
---

The cut-off is the significant marker because it directly corresponds to the definition of the AR and MA processes:

<br>

## For AR($p$) (Cut-off in PACF)
---

> An AR($p$) model states that the current value $Y_t$ is a linear combination of the $p$ previous values:

$$Y_t = c + \phi_1 Y_{t-1} + \phi_2 Y_{t-2} + \dots + \phi_p Y_{t-p} + \epsilon_t$$

The term $\phi_p$ is the direct link. For any lag $k > p$, there is no direct link between $Y_t$ and $Y_{t-k}$.

Since the PACF measures the direct link (with intermediate effects removed), the PACF must be zero for all lags $k > p$.

The last significant spike on the PACF determines $p$ (the order of the AR component).

<br>

## For MA($q$) (Cut-off in ACF)
---


> An MA($q$) model states that the current value $Y_t$ is a linear combination of the $q$ previous error terms (shocks):

$$Y_t = \mu + \epsilon_t + \theta_1 \epsilon_{t-1} + \theta_2 \epsilon_{t-2} + \dots + \theta_q \epsilon_{t-q}$$

The error term $\epsilon_t$ is only directly correlated with $\epsilon_{t-1}$ up to $\epsilon_{t-q}$.

- The ACF measures the total correlation (the "memory" imparted by these past shocks).
    - Because the process only "remembers" $q$ past shocks, the correlation between $Y_t$ and $Y_{t-q}$ is the last direct link.
    - For any lag $k > q$, there is no correlation.

- The last significant spike on the ACF determines $q$ (the order of the MA component).

<br>

## For both AR & MA
---

- If the plots show decay on both the ACF and PACF, the model is likely an ARMA($p, q$) with both components,
- Requiring a more complex fitting process to find the best $p$ and $q$.
